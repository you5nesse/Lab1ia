```python
import nltk
import spacy
import sklearn

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")

nlp = spacy.load("en_core_web_sm")
```

```python
import re

text = "The Students LOVE Natural Language Processing (NLP)!!! They were running to the library, because the courses are amazing... #AI @MIT :)"

# 1. Minuscules
text_lower = text.lower()

# 2. Suppression de la ponctuation et des caractères spéciaux
text_clean = re.sub(r"[^\w\s]", " ", text_lower)
text_clean = re.sub(r"\s+", " ", text_clean).strip()

print("Avant :", text)
print("Après :", text_clean)
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

tokens = word_tokenize(text_clean)
print(tokens)

stop_words = set(stopwords.words("english"))
print(sorted(stop_words))

tokens_filtrés = [t for t in tokens if t not in stop_words]
print(tokens_filtrés)

# Ajout d'un mot sans valeur informative pour ce texte
stop_words.add("amazing")
tokens_filtrés = [t for t in tokens if t not in stop_words]
print(tokens_filtrés)

# Proportion supprimée
prop = (len(tokens) - len(tokens_filtrés)) / len(tokens)
print(f"{len(tokens)} tokens -> {len(tokens_filtrés)} restants ({prop:.0%} supprimés)")
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

# Lemmatisation par défaut (considère chaque mot comme un nom)
tokens_lemmatisés = [lemmatizer.lemmatize(word) for word in tokens_filtrés]



print(tokens_lemmatisés)
```
